### `is.na()`
so basically, what na is.na does is to find if there are any missing values
NA = missing value

```r
x = NA   # assign NA to variable x
is.na(x)  # is it missing?
```

```r
x = c(11, NA, 13, NA)
is.na(x)
```

```r
x = c(11, NA, 13, NA)  # vector
mean(x)
```

```r
mean(x, na.rm = TRUE)  # NAs can be removed
```

```r
x = c(11, NA, 13, NA)
x
```

```r
which(is.na(x))
```

```r
x = c(11, NA, 13, NA)
x
```

```r
sum(is.na(x))
```

```r
x = c(11, NA, 13, NA)
x
```

```r
complete.cases(x)
```

### `if else()`

```r
x = 5
if (x == 3) { x = x - 1 } else { x = 2 * x }
x
```

```r
x = 3
if (x == 3) { x = x - 1 } else { x = 2 * x }
x
```

```r
x = 6
if (x > 3) {
  print("The value is more than 3")
} else {
  print("The value is less than 3")
}
```

```r
x = 2
if (x > 3) {
  print("The value is more than 3")
} else {
  print("The value is less than 3")
}
```

```r
x = 5
if (x == 3) {
  x = x - 1
} else if (x < 3) {
  x = x + 5
} else {
  x = 2 * x
}
x
```

```r
x = 2
if (x == 3) {
  x = x - 1
} else if (x < 3) {
  x = x + 5
} else {
  x = 2 * x
}
x
```

```r
x = 3
if (x == 3) {
  x = x - 1
} else if (x < 3) {
  x = x + 5
} else {
  x = 2 * x
}
x
```

### `which()`

```r
x = c(10, 15, 8, 14, 6, 12)
x
```

```r
which(x == 14)
```

```r
which(x != 12)
```

```r
which(x > 10)
```

```r
x = matrix(nrow = 3, ncol = 3, data = 1:9)
x
```

```r
which.min(x)  # find which is the minimum
```

```r
which.max(x)  # find which is the maximum
```

```r
x = matrix(nrow = 3, ncol = 3, data = 1:9)
x
```

```r
which(x %% 2 == 1)
```

```r
which(x %% 2 == 1, arr.ind = TRUE)   # gives row, column positions
```
