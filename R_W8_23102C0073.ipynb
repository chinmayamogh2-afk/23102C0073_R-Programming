### 1. Creating and merging lists

```r
# Create two lists
list1 <- list(1, 2, 3)
list2 <- list("water", "juice", "lemonade")

# Display the lists
list1
list2

# Merge the lists
list12 <- c(list1, list2)
list12
```

### 2. Converting a list to a vector — `unlist()`

```r
list1 <- list(1, 2, 3)
list2 <- list("water", "juice", "lemonade")

unlist(list1)
unlist(list2)

mode(list1)
mode(unlist(list1))
```

### 3. Appending to a list — `append()`

```r
list1 <- list(1, 2, 3)
list2 <- list("water", "juice", "lemonade")

append(list1, 100)
append(list2, "coffee")
```

### 4. Appending after a particular position

```r
list1 <- list(1, 2, 3)
list2 <- list("water", "juice", "lemonade")

append(list1, 100, after = 2)
append(list2, "coffee", after = 2)
```

### 5. Removing an element from a list

```r
list1 <- list(1, 2, 3)
list2 <- list("water", "juice", "lemonade")

list1[-2]
list2[-1]
```

### 6. Extracting elements from a list

```r
list1 <- list(1, 2, 3, 4, 5, 6)
list2 <- list("water", "juice", "lemonade", "tea", "coffee", "milk")

# Extract a range
list1[2:4]

# Extract selected positions
list1[c(1, 3, 5)]

# Same operations for list2
list2[2:4]
list2[c(1, 3, 5)]
```

### 7. Heterogeneous / mixed-mode lists

```r
# Lists can contain different types of values
ab <- list(1, 2, 3, "X", "Y", "Z")

# Give the list dimensions
dim(ab) <- c(2, 3)

print(ab)
mode(print(ab))
```



### 8. Conditional vector indexing

```r
x <- 1:10
x

# Values greater than 5
x[x > 5]

# Even values
x[x %% 2 == 0]

# Odd values
x[x %% 2 == 1]
```

### 9. Vector with missing observations

```r
x <- 1:10

# Introduce a missing value
x[5] <- NA
x

# Remove missing values
y <- x[!is.na(x)]
y

# Mean with NA and without NA
mean(x)
mean(y)
```

### 9. Negative integer *indexing*

```r
x <- 1:10
x

# Exclude positions 1 through 5
x[-(1:5)]

# Equivalent positive indexing
x[6:10]
```

```r
# Convert categorical values to a factor
factor(c("Red", "Green", "Green", "Blue", "Red"))
```

### 13. Factor with explicitly specified levels and labels

```r
y <- c(1, 4, 3, 5, 4, 2, 4)

possible.dieface <- c(1, 2, 3, 4, 5, 6)

labels.dieface <- c(
  "one", "two", "three", "four", "five", "six"
)

facy <- factor(
  y,
  levels = possible.dieface,
  labels = labels.dieface
)

facy
```

### 18. `class()`

### 19. `unclass()` with factors

```r
brands <- c("A", "A", "B", "B", "B", "B", "C")
brands

brands_fac <- factor(brands)
brands_fac

# Convert factor codes back to their underlying numbers
unclass(brands_fac)
```

### 20. Using factor codes to index another vector

```r
colours <- c("blue", "green", "red")
colours

# Use the underlying factor codes as indexes
colours[unclass(brands_fac)]
```

### 21. Inspecting factor codes

### 22. Controlling factor level order

```r
x <- factor(
  c("lemonade", "lemonade", "juice", "lemonade", "water"),
  levels = c("water", "juice", "lemonade")
)

x

unclass(x)
levels(x)
```

### 23. Ordered factors

```r
income <- ordered(
  c("high", "high", "low", "medium", "medium"),
  levels = c("low", "medium", "high")
)

income
unclass(income)
levels(income)
```

## 26. `format()` with width and justification

```r
x <- c("A", "BB", "CCC", "DDDD")

format(x, width = 7, justify = "centre")
format(x, width = 14, justify = "centre")

format(x, width = 7, justify = "left")
format(x, width = 7, justify = "right")
format(x, width = 7, justify = "none")
```
